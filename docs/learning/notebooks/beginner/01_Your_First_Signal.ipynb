# 📡 What is a Biosignal?

Your body is **electric**! Here's the amazing truth:

- Your heart creates electricity every time it beats

### Let's See the Simplest Signal: A Wave

Before we look at real body signals, let's see what a simple wave looks like:

```python
# Create a simple wave (like a sound wave or water wave)
time = np.linspace(0, 2, 1000)  # 2 seconds of time
frequency = 2  # How many waves per second - Try changing this!
wave = np.sin(2 * np.pi * frequency * time)

# Draw it!
plt.figure(figsize=(12, 4))
plt.plot(time, wave, linewidth=2, color='dodgerblue')
plt.xlabel('Time (seconds)', fontsize=12)
plt.ylabel('Signal Strength', fontsize=12)
plt.title('🌊 A Simple Wave Signal', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()

print(f"This wave repeats {frequency} times every second!")
```

# 💓 Let's See a Real Heartbeat!

Now for the exciting part! We're going to create a signal that looks just like what doctors see on heart monitors.

This is called an **ECG** (electrocardiogram). It shows the electrical activity of your heart.

## What Makes a Heartbeat?

Each heartbeat has a special pattern with peaks and valleys. The biggest peak is called the **R peak** - that's when your heart muscle squeezes!

```python
def create_heartbeat_signal(duration, heart_rate):
    """
    Create a realistic heartbeat signal!
    
    duration: how many seconds of signal to create
    heart_rate: beats per minute (60-180 is normal range)
    """
    sampling_rate = 500  # How many measurements per second
    time = np.linspace(0, duration, int(duration * sampling_rate))
    signal = np.zeros(len(time))
    
    beat_interval = 60 / heart_rate  # Time between heartbeats
    
    # Create each heartbeat
    for beat_time in np.arange(0, duration, beat_interval):
        # Find where this beat happens
        beat_index = int(beat_time * sampling_rate)
        
        # Create the heartbeat wave pattern
        beat_length = int(0.6 * sampling_rate)  # Each beat lasts 0.6 seconds
        if beat_index + beat_length < len(signal):
            t = np.linspace(0, 1, beat_length)
            
            # The PQRST complex (the heartbeat shape doctors look for)
            p_wave = 0.2 * np.exp(-((t - 0.2) ** 2) / 0.005)  # Small bump
            qrs_complex = 1.5 * np.exp(-((t - 0.35) ** 2) / 0.001)  # Big spike!
            t_wave = 0.3 * np.exp(-((t - 0.6) ** 2) / 0.01)  # Medium bump
            
            beat = p_wave + qrs_complex + t_wave
            signal[beat_index:beat_index + beat_length] += beat
    
    return time, signal

print("✅ Heartbeat generator ready!")
```

```python
# Create a heartbeat signal
heart_rate = 75  # Try changing this! (60-120 is a good range)
duration = 5  # seconds

time, ecg_signal = create_heartbeat_signal(duration, heart_rate)

# Plot it!
plt.figure(figsize=(14, 5))
plt.plot(time, ecg_signal, linewidth=2, color='crimson')
plt.xlabel('Time (seconds)', fontsize=12)
plt.ylabel('Voltage (mV)', fontsize=12)
plt.title(f'💓 Your Heartbeat at {heart_rate} BPM (Beats Per Minute)', 
          fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)

# Point out the peaks (heartbeats)
from scipy.signal import find_peaks
peaks, _ = find_peaks(ecg_signal, height=1.0, distance=100)
plt.plot(time[peaks], ecg_signal[peaks], 'go', markersize=10, label='Heartbeats!')
plt.legend(fontsize=11)
plt.show()

num_beats = len(peaks)
print(f"\n🫀 We detected {num_beats} heartbeats!")
print(f"📊 That's {num_beats / duration * 60:.0f} beats per minute")
print(f"\n💡 Each peak is one heartbeat - that's when your heart pumps blood!")
```

## 🎉 You Did It!

You just created and analyzed a heartbeat signal! The green dots show exactly when each heartbeat happens.

**Real-world use:** This is exactly what a heart monitor in a hospital shows. Doctors use this to check if your heart is healthy!

# 🎮 Make It Interactive!

Now let's have some fun! We'll create a simple controller where you can change the heart rate and watch the signal change in real-time.

This is like being a heart rate DJ! 🎧

```python
def plot_heart_rate(hr):
    """Create a heartbeat plot for the given heart rate"""
    time, signal = create_heartbeat_signal(5, hr)
    
    plt.figure(figsize=(14, 5))
    plt.plot(time, signal, linewidth=2, color='crimson')
    plt.xlabel('Time (seconds)', fontsize=12)
    plt.ylabel('Voltage (mV)', fontsize=12)
    
    # Add context about the heart rate
    if hr < 60:
        status = "Very Slow (Athlete or sleeping)" 
        color = "blue"
    elif hr <= 80:
        status = "Normal Resting"
        color = "green"
    elif hr <= 120:
        status = "Exercise or Excitement"
        color = "orange"
    else:
        status = "Intense Exercise!"
        color = "red"
    
    plt.title(f'💓 Heart Rate: {hr} BPM - {status}', 
              fontsize=14, fontweight='bold', color=color)
    plt.grid(True, alpha=0.3)
    plt.ylim(-0.2, 2.0)
    plt.show()
    
    # Count the beats
    peaks, _ = find_peaks(signal, height=1.0, distance=100)
    actual_bpm = len(peaks) / 5 * 60
    print(f"⏱️  Time between beats: {60/hr:.2f} seconds")
    print(f"🫀 Total beats in 5 seconds: {len(peaks)}")

# Try different heart rates!
print("Try these scenarios:\n")
print("🛌 Sleeping: 45 BPM")
print("😌 Relaxed: 65 BPM")
print("🚶 Walking: 90 BPM")
print("🏃 Running: 140 BPM")
print("\nChange the number below and run this cell again!\n")

plot_heart_rate(75)  # Try changing this number!
```

# 🧠 Brain Waves Preview

Your heart isn't the only part of you that makes electrical signals. Your **brain** does too!

Different brain waves appear when you're doing different things:

- 🧘 **Alpha waves (8-13 Hz)**: Relaxed, eyes closed, calm
- 🎯 **Beta waves (13-30 Hz)**: Alert, focused, problem-solving
- 😴 **Delta waves (0.5-4 Hz)**: Deep sleep
- 💭 **Theta waves (4-8 Hz)**: Drowsy, daydreaming

Let's see what these look like!

```python
def create_brain_wave(wave_type, duration=3):
    """
    Create different types of brain waves (EEG signals)
    
    wave_type: 'alpha', 'beta', 'theta', or 'delta'
    """
    sampling_rate = 256  # Measurements per second
    time = np.linspace(0, duration, int(duration * sampling_rate))
    
    # Different frequency ranges for different brain states
    wave_configs = {
        'delta': {'freq': 2, 'name': 'Delta (Deep Sleep)', 'emoji': '😴', 'color': 'navy'},
        'theta': {'freq': 6, 'name': 'Theta (Drowsy)', 'emoji': '💭', 'color': 'blue'},
        'alpha': {'freq': 10, 'name': 'Alpha (Relaxed)', 'emoji': '🧘', 'color': 'green'},
        'beta': {'freq': 20, 'name': 'Beta (Alert/Focused)', 'emoji': '🎯', 'color': 'orange'}
    }
    
    config = wave_configs[wave_type]
    
    # Create the brain wave with some natural variation
    signal = np.sin(2 * np.pi * config['freq'] * time)
    
    # Add realistic amplitude variation
    envelope = 1 + 0.3 * np.sin(2 * np.pi * 0.5 * time)
    signal = signal * envelope
    
    # Add a tiny bit of noise (your brain is never perfectly regular!)
    noise = np.random.normal(0, 0.1, len(signal))
    signal = signal + noise
    
    return time, signal, config

print("✅ Brain wave generator ready!")
```

## 🧘 Relaxed vs. Focused: Alpha vs. Beta Waves

Let's compare what your brain looks like when you're relaxed versus when you're concentrating hard!

```python
# Create both types of brain waves
time_alpha, alpha_signal, alpha_config = create_brain_wave('alpha')
time_beta, beta_signal, beta_config = create_brain_wave('beta')

# Plot them side by side
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8))

# Alpha waves (relaxed)
ax1.plot(time_alpha, alpha_signal, linewidth=2, color=alpha_config['color'])
ax1.set_ylabel('Brain Activity (µV)', fontsize=11)
ax1.set_title(f"{alpha_config['emoji']} {alpha_config['name']} - Eyes Closed, Peaceful", 
              fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.set_xlim(0, 3)

# Beta waves (focused)
ax2.plot(time_beta, beta_signal, linewidth=2, color=beta_config['color'])
ax2.set_xlabel('Time (seconds)', fontsize=11)
ax2.set_ylabel('Brain Activity (µV)', fontsize=11)
ax2.set_title(f"{beta_config['emoji']} {beta_config['name']} - Solving Math Problems", 
              fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.set_xlim(0, 3)

plt.tight_layout()
plt.show()

print("\n👀 Look closely! Can you see the difference?")
print(f"\n🧘 Alpha waves: {alpha_config['freq']} cycles per second (slower, bigger waves)")
print(f"🎯 Beta waves: {beta_config['freq']} cycles per second (faster, smaller waves)")
print("\n💡 Beta waves wiggle faster because your brain is more active!")
```

# 🏆 Bonus Challenge: Build a Mood Detector!

Ready for something cool? Let's combine heart rate and brain waves to create a simple "mood detector"!

## The Science:
- **Stressed**: Fast heart rate (100+ BPM) + Beta brain waves (alert)
- **Relaxed**: Slow heart rate (60 BPM) + Alpha brain waves (calm)
- **Sleeping**: Slow heart rate (55 BPM) + Delta brain waves (deep sleep)
- **Exercising**: Fast heart rate (140 BPM) + Beta brain waves (focused)

Let's build it!

```python
def detect_mood(heart_rate, brain_wave_type):
    """
    Simple mood detector based on biosignals!
    
    heart_rate: beats per minute
    brain_wave_type: 'delta', 'theta', 'alpha', or 'beta'
    """
    # Create the signals
    time_heart, ecg = create_heartbeat_signal(4, heart_rate)
    time_brain, eeg, brain_config = create_brain_wave(brain_wave_type, duration=4)
    
    # Mood detection logic
    if brain_wave_type == 'delta' and heart_rate < 60:
        mood = "Deep Sleep"
        emoji = "😴💤"
        color = "navy"
    elif brain_wave_type == 'alpha' and heart_rate < 75:
        mood = "Relaxed & Calm"
        emoji = "🧘‍♀️✨"
        color = "green"
    elif brain_wave_type == 'beta' and heart_rate > 100:
        mood = "Stressed or Excited"
        emoji = "😰⚡"
        color = "red"
    elif brain_wave_type == 'beta' and heart_rate > 130:
        mood = "Intense Exercise"
        emoji = "🏃‍♀️💪"
        color = "orange"
    elif brain_wave_type == 'theta':
        mood = "Drowsy or Daydreaming"
        emoji = "💭🌙"
        color = "purple"
    else:
        mood = "Normal Awake State"
        emoji = "😊👍"
        color = "blue"
    
    # Plot both signals
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8))
    
    # Heart signal
    ax1.plot(time_heart, ecg, linewidth=2, color='crimson')
    ax1.set_ylabel('ECG (mV)', fontsize=11)
    ax1.set_title(f'💓 Heart Rate: {heart_rate} BPM', fontsize=12, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    
    # Brain signal
    ax2.plot(time_brain, eeg, linewidth=2, color=brain_config['color'])
    ax2.set_xlabel('Time (seconds)', fontsize=11)
    ax2.set_ylabel('EEG (µV)', fontsize=11)
    ax2.set_title(f"🧠 Brain Waves: {brain_config['name']}", fontsize=12, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Show the mood detection result
    print("\n" + "="*50)
    print(f"🎯 MOOD DETECTION RESULT")
    print("="*50)
    print(f"\n{emoji} Detected Mood: {mood}")
    print(f"\n📊 Based on:")
    print(f"   • Heart Rate: {heart_rate} BPM")
    print(f"   • Brain Wave: {brain_config['name']}")
    print("\n" + "="*50)

print("✅ Mood detector ready!")
```
